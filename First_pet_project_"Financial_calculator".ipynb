```python
#Counting all the income for a month
while True:
  try:
    total_income = int(input('Total income:' ' '))
    if total_income < 0:
      print('Total income can not be less then 0')
    break
  except:
   print('Enter numeric values')

#counting fixed expenses
fixed_expenses = 0
print('Enter your fixed expenses one by one, then print done')

while True:
  user_fixed_expenses = input('Fixed expenses or "done:"')
  if user_fixed_expenses == 'done':
    print('ALL DONE')
    break
  try:
    fexpense = int(user_fixed_expenses)
    if fexpense < 0:
      print('Expenses can not be less than zero')
      continue
    fixed_expenses += fexpense
    print('Total fixed expenses:', fixed_expenses)
  except:
    print('Enter numeric values')

    #counting flexible expenses
flexible_expenses = 0
print('Enter your flexible expenses one by one, then print done')

while True:
  user_flexible_expenses = input('Flexible expenses or "idone:"')
  if user_flexible_expenses == 'idone':
    print('ALL DONE')
    break
  try:
    expense = int(user_flexible_expenses)
    if expense < 0:
      print('Expenses can not be less than zero')
      continue
    flexible_expenses += expense
    print('Total expenses:', flexible_expenses)
  except:
    print('Enter numeric values')

total_expenses = (fixed_expenses+flexible_expenses)
print('Total expenses:', total_expenses)

total_balance = (int(total_income) - int(total_expenses))
print('Remaining balance:', total_balance)

```
